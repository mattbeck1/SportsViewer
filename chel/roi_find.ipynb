*The pupose of this notebook is to find ROI values for the scorebug*

```python
import cv2 as cv
```

```python
image = cv.imread('../chel/chel.png')

# Set the maximum display dimensions
screen_width, screen_height = 800, 600  # Adjust as needed to fit your screen

# Get the current dimensions of the image
img_height, img_width = image.shape[:2]

# Calculate the scaling factor
scale = min(screen_width / img_width, screen_height / img_height)

# Resize the image if it's larger than the screen
if scale < 1:
    resized_image = cv.resize(image, (int(img_width * scale), int(img_height * scale)))
else:
    resized_image = image  # No resizing needed if image fits the screen

# Select ROI on the resized image
roi = cv.selectROI("Select ROI", resized_image)

# Adjust ROI coordinates back to original image scale
x, y, w, h = roi
x = int(x / scale)
y = int(y / scale)
w = int(w / scale)
h = int(h / scale)

print(f"Y : {y}\nX : {x}\nW : {w}\nH : {h}")

# Crop the ROI from the original image
roi_cropped = image[y:y+h, x:x+w]

# Display the cropped ROI for verification
cv.imshow("Cropped ROI", roi_cropped)
cv.waitKey(0)
cv.destroyAllWindows()
```
